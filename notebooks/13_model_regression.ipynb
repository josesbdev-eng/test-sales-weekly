## 1. Main variables: SpendChannel1, SpendChannel2, SpendCompetitors


```python
# Set target variable
var_target = "Conversions"

# Independent variables
DataX = (
    Data11
    >> select(_.SpendChannel1, _.SpendChannel2, _.SpendCompetitors)
)
X = DataX.values

# Target variable
DataY = Data11[f"""{var_target}"""]
Y = DataY.values

```

## Partitioning data (Train - Test)

Based on the time appeareance

```python
# Librerias
from sklearn.model_selection import train_test_split

# Particionar la data
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, shuffle=False)
```

## Multiple Linear Regression Model

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# 1. Entrenar la regresión lineal múltiple
modelo_lineal = LinearRegression()
modelo_lineal.fit(X_train, y_train)

# 2. Predecir sobre el conjunto de prueba
y_pred = modelo_lineal.predict(X_test)

# 3. Calcular las métricas
r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

n = len(y_test)          # Número de observaciones de prueba
p = X_test.shape[1]      # Número de variables predictoras

r2_ajustado = (
    1 - (1 - r2) * (n - 1) / (n - p - 1)
    if n > p + 1 else np.nan
)

resultados_1 = pd.DataFrame({
    "Modelo": ["LinearRegression-Main"],
    "Adjusted R-Squared": [r2_ajustado],
    "R-Squared": [r2],
    "RMSE": [rmse]
}).set_index("Modelo")

display(resultados_1)

# 4. Mostrar intercepto y coeficientes
print("Intercepto:", modelo_lineal.intercept_)

coeficientes = pd.DataFrame({
    "Variable": [f"X{i + 1}" for i in range(X_train.shape[1])],
    "Coeficiente": np.asarray(modelo_lineal.coef_).ravel()
})

display(coeficientes)
```

## Hyphotesis testing

To main plus time vars model.



```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

if isinstance(X_train, pd.DataFrame):
    X = X_train.copy().astype(float)
else:
    nombres_variables = [
        "Spend Channel 1",
        "Spend Channel 2",
        "Spend Competitors",
        "Year",
        "Month",
        "WeekNumber"
    ]

    X = pd.DataFrame(
        np.asarray(X_train, dtype=float),
        columns=nombres_variables
    )

y = np.asarray(y_train, dtype=float).reshape(-1)

if len(X) != len(y):
    raise ValueError("X_train e y_train tienen distinta cantidad de filas.")

if not np.isfinite(X.to_numpy()).all() or not np.isfinite(y).all():
    raise ValueError("Existen valores faltantes o infinitos.")

constantes = X.columns[X.nunique() <= 1].tolist()

if constantes:
    print("Variables constantes excluidas:", constantes)
    X = X.drop(columns=constantes)

X_ols = sm.add_constant(X, has_constant="add")

if len(X_ols) <= X_ols.shape[1]:
    raise ValueError("No hay suficientes observaciones para la inferencia.")

if np.linalg.matrix_rank(X_ols.to_numpy()) < X_ols.shape[1]:
    raise ValueError(
        "Hay colinealidad perfecta. Revisa variables redundantes."
    )

if np.std(y, ddof=1) == 0:
    raise ValueError("La variable objetivo no presenta variación.")

modelo_ols = sm.OLS(y, X_ols).fit()

print(modelo_ols.summary())
alpha = 0.05

prueba_global = pd.DataFrame({
    "Estadístico F": [modelo_ols.fvalue],
    "p-valor": [modelo_ols.f_pvalue],
    "Decisión": [
        "Rechazar H0"
        if modelo_ols.f_pvalue < alpha
        else "No rechazar H0"
    ]
})

print("\nPrueba global del modelo:")
display(prueba_global)

intervalos = modelo_ols.conf_int(alpha=alpha)

tabla_coeficientes = pd.DataFrame({
    "Coeficiente": modelo_ols.params,
    "Error estándar": modelo_ols.bse,
    "Estadístico t": modelo_ols.tvalues,
    "p-valor": modelo_ols.pvalues,
    "IC 95% inferior": intervalos[0],
    "IC 95% superior": intervalos[1]
})

tabla_coeficientes["Decisión (α=0.05)"] = np.where(
    tabla_coeficientes["p-valor"] < alpha,
    "Rechazar H0",
    "No rechazar H0"
)

tabla_coeficientes.index.name = "Variable"

print("\nPruebas individuales:")
display(tabla_coeficientes)
```

## Hypothesis Testing

A significance level of $\alpha = 0.05$ was used to evaluate.

### 1. Overall Model Significance: F-Test

The hypotheses are:

$$
H_0: \beta_1 = \beta_2 = \beta_3 = \beta_4 = \beta_5 = \beta_6 = 0
$$

$$
H_1: \text{At least one } \beta_j \neq 0, \quad j \in \{1,\ldots,6\}
$$

The results are:

$$
F = 220.27816, \qquad p\text{-value} = 1.297350 \times 10^{-72}
$$

Since the p-value is below $0.05$, we reject $H_0$. The model is statistically significant overall, providing evidence that at least one predictor is associated with conversions.

This result does not imply that every predictor is statistically significant.

<br />

### 2. Individual Coefficient Significance: t-Tests

For each predictor $X_j$, the hypotheses are:

$$
H_0: \beta_j = 0
$$

$$
H_1: \beta_j \neq 0
$$



| Predictor | Estimated coefficient | p-value | Conclusion at $\alpha = 0.05$ |
|---|---:|---:|---|
| Spend Channel 1 | 0.047208 | $5.123533 \times 10^{-65}$ | Reject $H_0$: significant positive association |
| Spend Channel 2 | 0.179978 | $1.724780 \times 10^{-39}$ | Reject $H_0$: significant positive association |
| Spend Competitors | -0.001043 | 0.8762772 | Do not reject $H_0$ |
| Year | -0.048321 | 0.7544142 | Do not reject $H_0$ |
| Month | -0.006206 | 0.9537544 | Do not reject $H_0$ |
| WeekNumber | 0.002387 | 0.9219199 | Do not reject $H_0$ |

### 3. Interpretation of the Coefficients

**Spend Channel 1:** A one-unit increase in investment is associated with an estimated increase of $0.047208$ units in conversions, holding the other predictors constant. Its 95% confidence interval, $[0.044046,\ 0.050369]$, excludes zero.

**Spend Channel 2:** A one-unit increase in investment is associated with an estimated increase of $0.179978$ units in conversions, holding the other predictors constant. Its 95% confidence interval, $[0.160171,\ 0.199785]$, also excludes zero.

These interpretations refer to the units and transformations used in the fitted model.

**Spend Competitors, Year, Month, and WeekNumber:** Their p-values exceed $0.05$, and their 95% confidence intervals include zero. Therefore, the results do not provide sufficient evidence of individual linear associations with conversions after controlling for the other predictors. This does not prove that their associations are zero.

**Intercept:** The intercept is statistically significant, with an estimated value of $3.041665$ and a p-value of $3.323560 \times 10^{-6}$. It represents the predicted conversion value when all predictors equal zero. Its practical interpretation may be limited if this combination is outside the observed data.
